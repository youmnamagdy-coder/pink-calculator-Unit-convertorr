```python
import tkinter as tk
from tkinter import ttk

# ---------- Colors ----------
BG_COLOR = "#FFF0F5"        # lavender blush
FRAME_COLOR = "#FFE4EC"     # soft pink
BUTTON_COLOR = "#FFC1D9"    # pastel pink
BUTTON_ACTIVE = "#FFA6C9"   # deeper pink on press
ACCENT_COLOR = "#FF8FAB"    # equals/convert button
TEXT_COLOR = "#5C4B51"      # soft brown-gray
BODY_COLOR = "#FF9EC0"      # calculator "body/case" color

# ---------- Calculator functions ----------
def calc_click(char):
    calc_display.insert(tk.END, char)

def calc_clear():
    calc_display.delete(0, tk.END)

def calc_equal():
    try:
        result = eval(calc_display.get())
        calc_display.delete(0, tk.END)
        calc_display.insert(tk.END, str(result))
    except Exception:
        calc_display.delete(0, tk.END)
        calc_display.insert(tk.END, "Error")

# ---------- Unit converter (6 options = 3 unit types x 2 directions) ----------
conversions = {
    "Length (m to ft)": lambda x: x * 3.28084,
    "Length (ft to m)": lambda x: x / 3.28084,
    "Weight (kg to lb)": lambda x: x * 2.20462,
    "Weight (lb to kg)": lambda x: x / 2.20462,
    "Temp (C to F)": lambda x: (x * 9 / 5) + 32,
    "Temp (F to C)": lambda x: (x - 32) * 5 / 9,
}

def convert():
    try:
        value = float(conv_input.get())
        choice = conv_choice.get()
        result = conversions[choice](value)
        conv_result.config(text=f"Result: {result:.4f}")
    except ValueError:
        conv_result.config(text="Enter a valid number")

def create_convert_button(parent):
    return tk.Button(parent, text="Convert", command=convert,
                      bg=ACCENT_COLOR, fg="white", activebackground=BUTTON_ACTIVE,
                      relief="raised", bd=4, font=("Arial", 12, "bold"))

# ---------- Main window ----------
root = tk.Tk()
root.title("Youmna Calculator")

WIN_W, WIN_H = 420, 620
screen_w = root.winfo_screenwidth()
screen_h = root.winfo_screenheight()
x = (screen_w // 2) - (WIN_W // 2)
y = (screen_h // 2) - (WIN_H // 2)
root.geometry(f"{WIN_W}x{WIN_H}+{x}+{y}")
root.configure(bg=BG_COLOR)

# Outer "case" frame -> gives the raised/3D body look
case_frame = tk.Frame(root, bg=BODY_COLOR, bd=8, relief="raised")
case_frame.pack(fill="both", expand=True, padx=14, pady=14)

# Brand name on top like a logo
tk.Label(case_frame, text="Youmna Calculator", font=("Georgia", 22, "bold"),
         bg=BODY_COLOR, fg="white").pack(pady=(16, 10))

style = ttk.Style()
style.theme_use("default")
style.configure("TNotebook", background=BODY_COLOR, borderwidth=0)
style.configure("TNotebook.Tab", background=FRAME_COLOR, foreground=TEXT_COLOR,
                font=("Arial", 12, "bold"), padding=[14, 8])
style.map("TNotebook.Tab", background=[("selected", BUTTON_COLOR)])
style.configure("TFrame", background=FRAME_COLOR)
style.configure("TCombobox", fieldbackground="white", background=FRAME_COLOR)

notebook = ttk.Notebook(case_frame)
notebook.pack(fill="both", expand=True, padx=10, pady=10)

# ----- Calculator tab -----
calc_frame = ttk.Frame(notebook)
notebook.add(calc_frame, text="Calculator")

calc_display = tk.Entry(calc_frame, font=("Arial", 22), justify="right",
                         bg="white", fg=TEXT_COLOR, insertbackground=TEXT_COLOR,
                         relief="sunken", bd=4)
calc_display.grid(row=0, column=0, columnspan=4, padx=10, pady=16, sticky="we")

buttons = [
    ('7', 1, 0), ('8', 1, 1), ('9', 1, 2), ('/', 1, 3),
    ('4', 2, 0), ('5', 2, 1), ('6', 2, 2), ('*', 2, 3),
    ('1', 3, 0), ('2', 3, 1), ('3', 3, 2), ('-', 3, 3),
    ('0', 4, 0), ('.', 4, 1), ('C', 4, 2), ('+', 4, 3),
]

for (text, row, col) in buttons:
    cmd = calc_clear if text == 'C' else (lambda t=text: calc_click(t))
    tk.Button(calc_frame, text=text, width=6, height=2,
              bg=BUTTON_COLOR, fg=TEXT_COLOR, activebackground=BUTTON_ACTIVE,
              relief="raised", bd=4, font=("Arial", 13, "bold"),
              command=cmd).grid(row=row, column=col, padx=4, pady=4)
    tk.Button(calc_frame, text="=", width=28, height=2,
bg=ACCENT_COLOR, fg="white", activebackground=BUTTON_ACTIVE,
          relief="raised", bd=4, font=("Arial", 14, "bold"),
          command=calc_equal).grid(row=5, column=0, columnspan=4, padx=4, pady=10)

# ----- Converter tab -----
conv_frame = ttk.Frame(notebook)
notebook.add(conv_frame, text="Unit Converter")

tk.Label(conv_frame, text="Enter value:", font=("Arial", 13), bg=FRAME_COLOR, fg=TEXT_COLOR).pack(pady=(24, 6))
conv_input = tk.Entry(conv_frame, font=("Arial", 15), bg="white", fg=TEXT_COLOR, relief="sunken", bd=3)
conv_input.pack(pady=6)

tk.Label(conv_frame, text="Choose conversion:", font=("Arial", 13), bg=FRAME_COLOR, fg=TEXT_COLOR).pack(pady=(18, 6))
conv_choice = ttk.Combobox(conv_frame, values=list(conversions.keys()), state="readonly", width=25)
conv_choice.current(0)
conv_choice.pack(pady=6)

create_convert_button(conv_frame).pack(pady=18)

conv_result = tk.Label(conv_frame, text="Result: ", font=("Arial", 15), bg=FRAME_COLOR, fg=TEXT_COLOR)
conv_result.pack(pady=12)

root.mainloop()
```
